# 1. Introduction to LangChain with Google Gemini

This notebook demonstrates how to get started with LangChain using Google Gemini (`langchain-google-genai`).

```python
# Load environment variables from .env
from dotenv import load_dotenv
import os

load_dotenv()
```

### 1. Initialize Gemini Model
`ChatGoogleGenerativeAI` automatically reads `GOOGLE_API_KEY` or `GEMINI_API_KEY` from your `.env` file.

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-1.5-flash",
    temperature=0.7
)
```

### 2. Simple Invocation
Send a prompt to the model and inspect the response.

```python
response = llm.invoke("Explain LangChain in simple words.")
print(response.content)
```

### 3. Using Messages (SystemMessage & HumanMessage)
Structure conversations using roles like system instructions and user input.

```python
from langchain_core.messages import SystemMessage, HumanMessage

messages = [
    SystemMessage(content="You are an expert AI coding assistant."),
    HumanMessage(content="Give me 3 key advantages of using LangChain.")
]

response = llm.invoke(messages)
print(response.content)
```

### 4. LCEL (LangChain Expression Language) with PromptTemplate
Chain prompt templates directly into Gemini using the `|` operator.

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful tutor explaining concepts clearly."),
    ("human", "Explain {concept} for a beginner.")
])

chain = prompt | llm | StrOutputParser()

result = chain.invoke({"concept": "Prompt Engineering"})
print(result)
```
